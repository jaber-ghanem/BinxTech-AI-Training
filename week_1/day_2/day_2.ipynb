```python
"The function here takes list of numbers and return the mean, max, and min in a dictionary"
from pydantic import BaseModel
class CalculationOutput(BaseModel):
    "This model describes the output of the calculation of a function over a list of numbers and calculates the mean, min, and max "
    mean:float
    min: float
    max:float 
     
def calculate_values(numbers:list[float])-> CalculationOutput | None:
    "This function provides the ability to calculate the minimum, maximum and mean for a list of float numbers"
    sum_numbers = 0   
    min_number=None
    max_number=None
    if len(numbers) ==0:
        print("No numbers provided")
        return None 
    for number in numbers:
        sum_numbers+=number
        if min_number is None :
            min_number = number
        if max_number is None :
            max_number= number
        if max_number <  number:
            max_number = number
        if min_number > number :
            min_number = number 
        
    mean = sum_numbers/len(numbers)
    max_number =max_number if max_number is not None else 0
    min_number = min_number if min_number is not None else 0
    output = CalculationOutput(mean=mean,max=max_number,min=min_number)
    return output
numbers_list:list[float] = [10,20,50,60,80 , -3 ,1,7,5 -86]
calculation_output = calculate_values(numbers=numbers_list)
if calculation_output is not None:
    print("Mean:" , calculation_output.mean)
    print("Max" , calculation_output.max)
    print("Min" , calculation_output.min)

        
    
```

```python
from pydantic import BaseModel
class Position(BaseModel):
    "Describes the position that the person is in"
    x:int = 0
    y:int = 0
class Person:
    _position_x:int
    _position_y:int 
    def __init__(self):
        self._position_x=0
        self._position_y=0
    def change_position(self, x:int | None,y:int | None):
        if x is not None:
            self._position_x+=x
        if y is not None:
            self._position_y+=y
    def get_postion(self)-> Position:
            return Position(x=self._position_x,y=self._position_y)
    def reset_position(self)-> None:
        self._position_x= 0
        self._position_y=0
person = Person()
person.change_position(x=10,y=5)
print(person.get_postion())
```
