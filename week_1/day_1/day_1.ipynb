```python
"This module is meant to extract the packages and their versions from the requirements file"
try:
    with open("../../requirements.txt" ,encoding="utf-8") as f:
        file_content = f.read()
        lines = file_content.split("\n")
        for line in lines:
            package_and_version = line.split("==")
            if len(package_and_version) == 2:
                package = package_and_version[0]
                version = package_and_version[1]
                print(package,version)
except FileNotFoundError as e:
    print(e)
```
