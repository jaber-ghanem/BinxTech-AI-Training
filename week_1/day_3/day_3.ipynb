```python
"Day 2"
# Here i create a 4x4 array by using linspace where the numbers start at 1 and end at 16 with 16 even seperations and then reshape it to be a 2d matrix with 4 rows and 4 columns
import numpy as np
array = np.linspace(1,16,16).reshape(4,4)
print(array)
```

Task 2:

```python
# Here i exract the second column of the last row, -1 indicates the first row from the opposite side (last one) and 1 is the second element
second_column_of_last_row= array[-1,1]
print(second_column_of_last_row)
```

Task 3:

```python
#Here i use boolean masking to find the values that are greater than the mean value of the array
mean_value = array.mean()
values_greater_than_mean = array[array>mean_value]
print(f"Mean: {mean_value}",values_greater_than_mean)
```

```python
# Create an array of 10s by creating an array of 1s and multiplying it with 10
array_of_10s = np.ones((1,4)) * 10
# Add 10 to every element
broadcasted_array=array+array_of_10s
print(broadcasted_array)
```
